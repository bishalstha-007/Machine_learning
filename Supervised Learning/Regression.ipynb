Regression models predicts continuous numerical values(eg: housing prices).

```python
# Topic: Linear & Polynomial Regression

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

# 1. Generate Systhetic Data
X = np.random.rand(100, 1) *  6 - 3
y = 0.5 *X**2 + X + 2 + np.random.randn(100, 1)

# 2. Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state= 42)

# 3. Transform for Polynomial Regression (Degree 2)
poly_features = PolynomialFeatures(degree=2, include_bias=False)
X_poly_train = poly_features.fit_transform(X_train)
X_poly_test = poly_features.transform(X_test)


# 4. Train Model
model = LinearRegression()
model.fit(X_poly_train, y_train)

# 5. Evaluate
predictions = model.predict(X_poly_test)
mse = mean_squared_error(y_test, predictions)
print(f"Polynomial Regression MSE: {mse:.4f}")
```
