Classification models separate data into distinct categories or classes (eg: spam vs ham).

```python
# Topics: Logistic Regresion, Support Vector Machines (SVM), and Random Forest

from sklearn.datasets import load_iris 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# 1. Load Dataset
data = load_iris()
X, y = data.data, data.target
X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.3, random_state=42)

# 2. Logistic Regression
lr_model = LogisticRegression(max_iter=200)
lr_model.fit(X_train, y_train)
print(f"Logistic Rgression Accuracy: {accuracy_score(y_test, lr_model.predict(X_test)):.4f}")

# 3. Support Vecotr Machine (SVM)
svm_model = SVC(kernel='rbf', C=1.0)
svm_model.fit(X_train, y_train)
print(f"SVM Accuracy : {accuracy_score(y_test, svm_model.predict(X_test)):.4f}")

# 4. Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
print(f"Random Forest Accuracy: {accuracy_score(y_test, rf_model.predict(X_test)):.4f}")
```
